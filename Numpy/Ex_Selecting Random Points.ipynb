```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import seaborn; seaborn.set_theme()
```

```python
mean=[0,0]
cov=[[1,2],
     [2,5]]

X = np.random.multivariate_normal(mean,cov,100)

ind = np.random.choice(X.shape[0],20)
sample = X[ind]
print(f"X shape {X.shape}\nSample shape {sample.shape}")
```

```python
plt.scatter(X[:,0],X[:,1], alpha=0.3) 
plt.scatter(sample[:,0],sample[:,1],edgecolors='black',facecolor='none', s=200)

plt.title('Taking sample from data')
plt.show()
```
