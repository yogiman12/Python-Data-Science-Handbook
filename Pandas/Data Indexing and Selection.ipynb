###  Data Selection in Series

####  Series as dictionary


```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],index=['a', 'b', 'c', 'd'])
data
```

```python
data['a']
```

```python
'a' in data 
```

```python
data.keys()
```

```python
data.values
```

```python
list(data.items())
```

```python
data['e'] = 99
data
```

####  Series as one-dimensional array


```python
data['a':'c']
```

```python
data[0:2]
```

```python
data[0]
```

```python
data[data>0.5]
```

```python
mask = (data>0.3) & (data<1)
data[mask]
```

```python
data[['a','c']]
```
