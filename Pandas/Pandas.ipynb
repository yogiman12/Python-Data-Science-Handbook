##### Constructing DataFrame objects

**A Pandas DataFrame can be constructed in a variety of ways. Here we’ll give several
 examples.**

#### From a single Series object. 

```python
pd.DataFrame(population,columns=['population'])
```

#####  From a list of dicts.  

```python
dicts = [{"a":i,"b":i*2 ,'c': i *3} for i in range(5)]
pd.DataFrame(dicts)
```

**Even if some keys in the dictionary are missing, Pandas will fill them in with NaN (i.e.,
 “not a number”)**

```python
pd.DataFrame([{'a': 1, 'b': 2}, {'b': 3, 'c': 4}])
```

###  The Pandas Index Object

```python
ind = pd.Index([1,2,3,4,5])
ind
```

####  Index as immutable array

```python
print(ind[1],ind[2:],ind[::2])
```

```python
print(ind.size, ind.shape, ind.ndim, ind.dtype)
```

**One difference between Index objects and NumPy arrays is that indices are immuta
ble—that is, they cannot be modified via the normal means:**

```python
ind[1] = 100
```

####  Index as ordered set

```python
indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])
```

```python
indA & indB # wrong
```

```python
indA.intersection(indB) # correct
```

```python
indA | indB # wrong
```

```python
indA.union(indB) # correct
```

```python
indA ^ indB # wrong
```

```python
indA.difference(indB) # correct
```
